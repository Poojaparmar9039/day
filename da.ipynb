```python
#1longest element in array

a=[1,2,2,34,56,6,4,3]
l=a[0]

for i in range(len(a)):
    if a[i]>l:
        l=a[i]

print(l)
```

```python
#2 second largest in array

a=[1,2,3,54,32,2,3]

l=a[0]
s=a[0]
for i in range(len(a)):
    if a[i]>l:
        s=l
        l=a[i]

    elif a[i]>s and a[i]!=l:
        s=a[i]

print("largest ", l)
print("second largest", s)
```

```python
#5 check array is sorted or not 

a=[1,2,3,4,5,16]

flag=0

for i in range(1,len(a)):
    if a[i]<a[i-1]:
        flag=1
        break

if flag==0:
    print("sorted")
else:
    print("unsorted")
```
