```python
#2 second largest in array

a=[1,2,3,54,32,2,3]

l=a[0]
s=a[0]
for i in range(len(a)):
    if a[i]>l:
        s=l
        l=a[i]

    elif a[i]>s and a[i]!=l:
        s=a[i]

print("largest ", l)
print("second largest", s)
```

```python
#5 check array is sorted or not 

a=[1,2,3,4,5,16]

flag=0

for i in range(1,len(a)):
    if a[i]<a[i-1]:
        flag=1
        break

if flag==0:
    print("sorted")
else:
    print("unsorted")
```

```python
#8 left rotate array by d places

a=[1,2,3,4,5,6,7,8,9]
n=len(a)
temp=[]
d=3

for i in range(d):
    temp.append(a[i])

for i  in range(d,n):
    a[i-d]=a[i]

for i in range(n-d,n):
    a[i]=temp[i-(n-d)]
    # print(temp[i-(n-d)],  i-(n-d),  i , n, d, n-d)basically i=n-d which cancel n-d

print(a)
```

```python
#9 right rotate the array by one place

a=[1,2,3,4,5,6]
n=len(a)
temp=a[n-1]

for i in range(n-1,0,-1):
    a[i]=a[i-1]

a[0]=temp
print(a)
```

```python
#10 right rotate the array by d places

a=[1,2,3,4,5,6,7,8,9]
n=len(a)
temp=[]
d=3

for i in range(n-d+1,n+1):
    temp.append(i)
print(temp)
for i in range(n-d-1,-1,-1):
    a[i+d]=a[i]

for i in range(d):
    a[i]=temp[i]

print(a)
```

```python
#11 move all zeros to end of the array
a=[1,0,2,3,0,0,4,5]
n=len(a)
temp=[]

for i in range(n):
    if a[i]!=0:
        temp.append(a[i])
m=len(temp)

for i in range(m):
    a[i]=temp[i]


for i in range(m,n):
    a[i]=0

print(a)
```

```python
#13 union of two array

a=[1,2,3,4,5,6]
b=[2,3,4,7,8]

n=len(a)
m=len(b)

i=0
j=0
union=[]

while i<n and j<m:
    if a[i]<b[j]:
        if not union or union[-1]!=a[i]:
            union.append(a[i])
        i+=1

    elif a[i]>b[j]:
        if not union or union[-1]!=b[j]:
            union.append(b[j])
        j+=1
    else:
        if not union or union[-1]!=a[i]:
            union.append(a[i])
        i+=1
        j+=1

while i<n:
    if union[-1]!=a[i]:
        union.append(a[i])
    i+=1

while j<m:
    if union[-1]!=b[j]:
        union.append(b[j])
    j+=1   
print(union)


```

```python
#14 intersection of two array

a=[1,2,3,4,5,6]
b=[2,3,4,7,8]

n=len(a)
m=len(b)

i=0
j=0
inter=[]


while i<n and j <m:
    if a[i]<b[j]:
        i+=1
    elif b[j]<a[i]:
        j+=1
    else:
        if not inter or inter[-1]!=a[i]:
            inter.append(a[i])

        i+=1
        j+=1

print(inter)
```
