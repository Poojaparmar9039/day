```python
arr = [12, 5, 78, 54, 3, 34]

l = float("-inf")
sl = float("-inf")
ssl = float("-inf")

for i in range(len(arr)):

    if arr[i] > l:
        ssl = sl
        sl = l
        l = arr[i]

    elif arr[i] > sl:
        ssl = sl
        sl = arr[i]

    elif arr[i] > ssl:
        ssl = arr[i]

print("Largest:", l)
print("Second largest:", sl)
print("Third largest:", ssl)
```

```python
#
arr1 = [1, 3, 5]
arr2 = [2, 4, 6]

arr3 = []

i = 0
j = 0

while i < len(arr1) and j < len(arr2):
    if arr1[i] < arr2[j]:
        arr3.append(arr1[i])
        i += 1
    else:
        arr3.append(arr2[j])
        j += 1

while i < len(arr1):
    arr3.append(arr1[i])
    i += 1

while j < len(arr2):
    arr3.append(arr2[j])
    j += 1

print(arr3)
```
