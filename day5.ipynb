```python
arr = [12, 5, 78, 54, 3, 34]

l = float("-inf")
sl = float("-inf")
ssl = float("-inf")

for i in range(len(arr)):

    if arr[i] > l:
        ssl = sl
        sl = l
        l = arr[i]

    elif arr[i] > sl:
        ssl = sl
        sl = arr[i]

    elif arr[i] > ssl:
        ssl = arr[i]

print("Largest:", l)
print("Second largest:", sl)
print("Third largest:", ssl)
```
