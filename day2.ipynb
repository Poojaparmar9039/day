```python
#Hasing
arr = [1, 5, 2, 1, 3, 1, 2]

n = len(arr)
has_arr = [0] * (max(arr) + 1)

for i in range(n):
    has_arr[arr[i]] += 1

print(has_arr)
```

```python
#49


num = int(input("Enter the number: "))

num = num + 1

while True:
    flag = True

    if num <= 1:
        flag = False
    else:
        for j in range(2, num):
            if num % j == 0:
                flag = False
                break

    if flag == True:
        print("Next prime number:", num)
        break

    num = num + 1
```
