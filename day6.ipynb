```python
#1
a=[1,2,3,4,5]
k=8

for i in range(len(a)):
    for j in range(len(a)):
        sum=a[i]+a[j]

    if sum==k:
        print(a[i])
        print(a[j])
        break

```

```python
#3

a=[1,2,3,4,5]
s=8

for i in range(len(a)):
    for j in range(len(a)):
        for k in range(len(a)):
           
            sum=a[i]+a[j]+a[k]

        if sum==s:
            print("i- ",a[i],"j- ",a[j],"k- ",a[k])
            break
```

```python
#4
a = [1, 2, 3, 4, 5]
sum = float("-inf")

for i in range(len(a)):
    for j in range(i+1, len(a)):
        k = a[i] + a[j]

        if sum < k:
            sum = k

print(sum)
```
