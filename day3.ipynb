```python
#5
num=int(input())
l=float("-inf")
s=float("inf")

while num>0:
    digit=num%10
    if digit>l:
        l=digit
    elif digit<s:
        s=digit
    num=num//10

print(l-s)
```

```python
#16
num = int(input())
temp = int(input())

result = 0
place = 1

while num > 0:
    digit = num % 10

    if digit != temp:
        result = result + digit * place
        place = place * 10

    num = num // 10

print(result)
```

```python
#17

num = int(input())
old = int(input())
new = int(input())

result = 0
place = 1

while num > 0:
    digit = num % 10

    if digit == old:
        digit = new

    result = result + digit * place
    place = place * 10

    num = num // 10

print(result)
```
